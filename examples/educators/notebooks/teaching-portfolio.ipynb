The cell below reads the checked-in FRED snapshot for `FEDFUNDS` and prints the latest observation on or before 2026-10-30. The six modules live under `docs/guides/teaching/`.


```python
import csv
from pathlib import Path

path = Path("data/teaching/snapshots/fred/FEDFUNDS.csv")
rows = []
with path.open(encoding="utf-8") as handle:
    for row in csv.DictReader(handle):
        raw = (row.get("FEDFUNDS") or "").strip()
        observed = (row.get("observation_date") or "").strip()
        if observed and raw and raw != "." and observed <= "2026-10-30":
            rows.append((observed, raw))
print(rows[-1])
```
